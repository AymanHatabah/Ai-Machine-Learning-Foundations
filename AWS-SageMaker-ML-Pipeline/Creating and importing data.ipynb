# 1. Importing Data & Environment Setup

In this section, we prepare the environment by importing the necessary standard Python libraries and downloading the dataset.

### Overview of Used Libraries:
* **`warnings`**: Used to suppress non-critical system warnings, keeping the notebook clean and readable.
* **`requests`**: A library used to send HTTP GET requests to download files and datasets directly via URLs over the internet.
* **`zipfile`**: Provides tools to read, write, and extract compressed `.zip` files directly in Python.
* **`io` (`BytesIO`)**: Allows us to handle binary data (the downloaded zip file) entirely in-memory as a file-like object, avoiding the need to write temporary files to the disk.
* **`pandas`**: The core Data Science library used for data manipulation, structured data cleaning, and creating DataFrames.
* **`scipy.io.arff`**: A specialized module from SciPy used to read `.arff` (Attribute-Relation File Format) data files, which is a common format in the UCI Machine Learning Repository.

```python
# Import core standard libraries
import warnings
import requests
import zipfile
import io

# Import data manipulation and dataset-specific reading tools
import pandas as pd
from scipy.io import arff

# Ignore non-critical warning messages to keep notebook output tidy
warnings.simplefilter('ignore')
```

```python
# URL pointing to the Vertebral Column dataset zip file from UCI ML Repository
f_zip = 'http://archive.ics.uci.edu/ml/machine-learning-databases/00212/vertebral_column_data.zip'

# Send an HTTP GET request to stream the remote file
r = requests.get(f_zip, stream=True)

# Read the streamed binary content into memory and parse it as a Zip file
vertebral_zip = zipfile.ZipFile(io.BytesIO(r.content))

# Extract all contents of the zip file into the current working directory
vertebral_zip.extractall()
```

### Step 3: Loading `.arff` Data into Pandas DataFrame

After extracting the raw archive, we process the dataset files into a structured tabular format:

* **`arff.loadarff(...)`**: Reads and parses the `.arff` (Attribute-Relation File Format) file. It returns a tuple containing:
  1. **Data Array (`data[0]`)**: A NumPy record array containing all rows and numerical features.
  2. **Metadata (`data[1]`)**: Information about attribute names, types, and target classes.
* **`pd.DataFrame(data[0])`**: Converts the extracted raw data array into a standard 2D Pandas DataFrame for exploratory data analysis (EDA).
* **`df.head()`**: Displays the first 5 rows of the DataFrame to inspect feature column names, data types, and initial values.

```python
# Load the 2-class classification dataset file from disk
data = arff.loadarff('data/column_2C_weka.arff')

# Convert the loaded ARFF structured array into a Pandas DataFrame
df = pd.DataFrame(data[0])

# Inspect the top 5 rows of the newly created DataFrame
df.head()
```
