### Introduction
Inside a Python class, we can write three different types of methods. The difference between them is what data they can access:

1. **Regular Methods (Instance Methods):** They automatically take `self` as the first parameter. They change or read data belonging to a specific object.
2. **Class Methods:** They take `cls` as the first parameter and use the `@classmethod` decorator. They work with the class variables and can change things for the whole class.
3. **Static Methods:** They do not take `self` or `cls`. They work like regular functions but live inside the class because they are logically connected to it.
---------------------

### Understanding our AI Instance Variables
To make sure you don't get lost, here is what the variables inside our AI class mean in the real world:

1. **`model_name`**: This is the specific AI architecture we are using (like `ResNet50` or `VGG16`). These are famous, powerful models used in Computer Vision to help computers understand images.
2. **`framework`**: This is the deep learning library we use to build and train the AI (like `PyTorch` or `TensorFlow`). `PyTorch` is highly popular on Kaggle because it is very flexible and easy to debug.
3. **`learning_rate` (LR)**: This is one of the most important numbers (Hyperparameters) in AI. It controls **how fast or slow** the model learns from its mistakes:
   - **High LR:** The model takes giant steps. It runs fast but might jump over the correct answer and fail.
   - **Low LR:** The model takes tiny baby steps. It is very precise but takes a long time (hours or days) to train.
   - **Good LR (like 0.001):** Balanced steps to reach the best accuracy efficiently.

```python
class AIModel:
    """
    A class to manage AI Model configurations.
    """
    # Class Variables
    total_models_created = 0

    def __init__(self, model_name: str, framework: str, learning_rate: float):
        # Instance Variables
        self.model_name = model_name
        self.framework = framework
        self.learning_rate = learning_rate
        
        # Update the class variable counter
        AIModel.total_models_created += 1

    # 1. Regular Instance Method
    def get_details(self):
        """Returns specific details about this model object."""
        return f"Model Name: {self.model_name} | Framework: {self.framework} | LR: {self.learning_rate}"

    # 2. Class Method (Used as an Alternative Constructor)
    @classmethod
    def from_string(cls, config_str: str):
        """
        Parses a string like 'ResNet50-PyTorch-0.001' 
        and returns a completely new AIModel object.
        """
        model_name, framework, lr_string = config_str.split("-")
        learning_rate = float(lr_string)
        
        # 'cls' represents the AIModel class itself.
        # This creates and returns the new instance automatically.
        return cls(model_name, framework, learning_rate)

    # 3. Static Method (Utility/Helper Function)
    @staticmethod
    def is_learning_rate_safe(lr: float):
        """
        Checks if a learning rate value is reasonable.
        It does not need any class or instance data to do this check.
        """
        if lr <= 0 or lr > 1.0:
            return False
        return True
```

### Testing Regular Methods (Instance Method) and Alternative Constructors (Class Method)

In real data pipelines, model settings often come as a single string from a text file or log. Instead of splitting the text manually in our main code, we use our class method `from_string` to create the object cleanly in one step.

```python
# Create model_1 using the regular constructor
model_1 = AIModel("VGG16", "TensorFlow", 0.01)
print("--- Model 1 Details (Regular) ---")
print(model_1.get_details())

# Create model_2 using a raw string and our @classmethod constructor
raw_config_data = "ResNet50-PyTorch-0.001"
model_2 = AIModel.from_string(raw_config_data)

print("\n--- Model 2 Details (From Class Method) ---")
print(model_2.get_details())

# Verify that the global counter updated correctly
print(f"\nTotal AI models in system: {AIModel.total_models_created}")
```

### Testing Static Methods

Static methods are great for simple checks and math operations. They do not accept `self` or `cls`, so we can run them directly using the class name without creating any objects first.

```python
# Testing different learning rates using the static method
lr_test_1 = 0.05
lr_test_2 = 5.5

print("--- Learning Rate Verification ---")
print(f"Is {lr_test_1} a safe learning rate? {AIModel.is_learning_rate_safe(lr_test_1)}")
print(f"Is {lr_test_2} a safe learning rate? {AIModel.is_learning_rate_safe(lr_test_2)}")
```

### Summary & Key Takeaways

In this notebook, I learned that:
1. **Regular methods** need an instance (`self`) to operate.
2. **Class methods** (`@classmethod`) are excellent for creating alternative constructors to build objects from external data formats like strings.
3. **Static methods** (`@staticmethod`) act as independent helper functions that belong to the class namespace for clean code organization.
