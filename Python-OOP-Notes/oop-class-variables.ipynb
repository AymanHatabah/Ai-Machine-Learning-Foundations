### What are Class Variables?
In the first notebook, we learned about **Instance Variables**, which hold data that is unique to each object (like a user's name). 

In this notebook, we explore **Class Variables**. These are variables that are **shared** among all instances of a class. This means the data is exactly the same for every object we create from this class.

```python
class Employee:
    """
    A class to represent a company employee.
    
    Class Variables:
    raise_amount (float): The salary increase rate for all employees.
    num_of_emps (int): A counter for the total number of employees.
    """
    
    # 1. Class Variables (Shared by all objects)
    raise_amount = 1.04  # 4% raise for everyone
    num_of_emps = 0      # Global counter starts at 0

    def __init__(self, first: str, last: str, pay: int):
        """
        Initializes a new employee instance.
        """
        # 2. Instance Variables (Unique for each object)
        self.first = first
        self.last = last
        self.pay = pay

        # We use the Class name here to increase the global counter centraly
        Employee.num_of_emps += 1

    def apply_raise(self):
        """
        Applies the raise amount to the employee's salary.
        """
        # We use 'self.raise_amount' to keep the code flexible
        self.pay = int(self.pay * self.raise_amount)
```

### Understanding Namespaces (How Python looks for variables)

When you call `object.variable`, Python searches in a specific order:
1. It looks inside the **Instance** memory first.
2. If it does not find the variable, it looks inside the **Class** memory.

Let's test this in the code below using `__dict__`, which shows what is saved inside the memory.

```python
# Check employee count before creating objects
print(f"Total employees before: {Employee.num_of_emps}")

# Create two employee objects
emp_1 = Employee("Ayman", "Hatabah", 5000)
emp_2 = Employee("John", "Smith", 6000)

print(f"Total employees now: {Employee.num_of_emps}")

# --- Inspecting Memory ---
print("\n--- Inside emp_1 Namespace ---")
print(emp_1.__dict__)  # Contains only first, last, and pay

print("\n--- Inside Employee Class Namespace ---")
print(Employee.__dict__)  # Contains raise_amount and num_of_emps

# --- Modifying the Class Variable ---
# Case A: Giving one specific employee a unique raise amount
emp_1.raise_amount = 1.10  # Ayman gets 10%

# Case B: Changing the raise rate for the rest of the company
Employee.raise_amount = 1.05  # Standard raise becomes 5%

# Apply raises and print the result
emp_1.apply_raise()
emp_2.apply_raise()

print("\n--- New Salaries ---")
print(f"{emp_1.first}'s new salary: {emp_1.pay} (10% raise applied)")
print(f"{emp_2.first}'s new salary: {emp_2.pay} (5% raise applied)")
```

### Real-World Example: AI Model Configuration

In Machine Learning and AI projects, we often use class variables to store global settings. For example, we want all our AI models to use the same hardware (like CPU or GPU) and the same random seed to make our results reproducible. This keeps our AI pipeline clean and consistent.

```python
class AIModelConfig:
    """
    A class to hold global configurations for an AI project.
    
    Class Variables:
    DEVICE (str): The hardware used for training (e.g., 'cpu' or 'cuda' for GPU).
    GLOBAL_SEED (int): The random seed to ensure reproducible results.
    FRAMEWORK (str): The main deep learning framework used.
    """
    # Class Variables (Global AI Settings)
    DEVICE = "cuda"          # Use GPU for all models
    GLOBAL_SEED = 42         # Seed for consistent numbers
    FRAMEWORK = "PyTorch"    # Main library

# We can access these settings directly anywhere without creating an object
print("--- Global AI Pipeline Settings ---")
print(f"Deep Learning Framework: {AIModelConfig.FRAMEWORK}")
print(f"Running all models on: {AIModelConfig.DEVICE}")
print(f"Global Random Seed: {AIModelConfig.GLOBAL_SEED}")
```

### Summary & Key Takeaways

In this part, I have learned that:
1. **Class Variables** are perfect for data that should be constant across all instances (like configurations or global counters).
2. Using `self.raise_amount` inside a method allows us to **override** the class variable for one specific object if we want to.
3. Class variables help us write clean, memory-efficient Python code.
