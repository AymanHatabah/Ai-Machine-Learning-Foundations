### What is Inheritance?
Inheritance allows us to create a new class that reuse, modify, and extend the behavior of an existing class. 
**Parent Class (Base Class):** The original class with shared attributes and methods.
**Child Class (Subclass):** The new class that inherits everything from the parent and adds its own specific features.

This follows the **DRY (Don't Repeat Yourself)** principle. Instead of writing the same basic AI configuration code over and over for different tasks, we create specialized subclasses!

```python
# 1. The Parent Class (Redefined from Part 3 so this notebook can run independently)
class AIModel:
    def __init__(self, model_name: str, framework: str, learning_rate: float):
        self.model_name = model_name
        self.framework = framework
        self.learning_rate = learning_rate

    def get_details(self):
        """Returns standard configuration details."""
        return f"Model: {self.model_name} | Framework: {self.framework} | LR: {self.learning_rate}"


# 2. Creating a Child Class for Computer Vision tasks
class VisionModel(AIModel):
    """
    A specialized subclass for Computer Vision models that process images.
    It inherits core properties from AIModel and adds image size configurations.
    """
    def __init__(self, model_name: str, framework: str, learning_rate: float, image_size: int):
        # super() automatically calls the __init__ of the parent class (AIModel)
        super().__init__(model_name, framework, learning_rate)
        
        # New attribute specific only to Computer Vision models
        self.image_size = image_size  # Example: 224 for 224x224 pixels

    # Method Overriding: Changing how get_details works for this subclass
    def get_details(self):
        """
        Overrides the parent method to include image size details for vision models.
        """
        # We reuse the parent's string and add our specific text to it
        ## (Dry) Method Overriding
        parent_text = super().get_details()
        return f"{parent_text} | Image Size: {self.image_size}x{self.image_size}"
```

### Adding another Subclass for Natural Language Processing (NLP)

Just like we created a subclass for images, we can create another subclass for text processing. An NLP model does not care about image pixels, but it needs to know the maximum number of words or tokens it can read at one time (`max_length`).

```python
# 3. Creating a Child Class for NLP tasks
class NLPModel(AIModel):
    """
    A specialized subclass for Natural Language Processing models that process text.
    
    It inherits core properties from AIModel and adds max token length configurations.
    """
    def __init__(self, model_name: str, framework: str, learning_rate: float, max_length: int):
        super().__init__(model_name, framework, learning_rate)
        
        # New attribute specific only to NLP models
        self.max_length = max_length  # Example: 512 tokens

    def get_details(self):
        """
        Overrides the parent method to include maximum text length details for NLP models.
        """
        parent_text = super().get_details()
        return f"{parent_text} | Max Length: {self.max_length} tokens"
```

### Testing Our Specialized Subclasses

Let's verify that our child classes correctly inherit the basic properties from `AIModel` while keeping their unique settings. 

We will also look at two helpful built-in Python tools:
- `isinstance(object, class)`: Checks if an object belongs to a specific class.
- `issubclass(child, parent)`: Checks if a class inherits from another class.

```python
# Create an instance of VisionModel
resnet_model = VisionModel("ResNet50", "PyTorch", 0.001, 224)

# Create an instance of NLPModel
bert_model = NLPModel("BERT-Base", "HuggingFace", 2e-5, 512)

print("--- Testing Child Class Methods ---")
print(resnet_model.get_details())
print(bert_model.get_details())

print("\n ---Type and Inheritance Verification Checks ---")
# Check if resnet_model belongs to VisionModel and AIModel
print(f"Is resnet_model object from a VisionModel? {isinstance(resnet_model, VisionModel)}")
print(f"Is resnet_model object from a an AIModel? {isinstance(resnet_model, AIModel)}")

# Check class hierarchies
print(f"Is NLPModel a subclass of AIModel? {issubclass(NLPModel, AIModel)}")
print(f"Is AIModel a subclass of VisionModel? {issubclass(AIModel, VisionModel)}")
```
