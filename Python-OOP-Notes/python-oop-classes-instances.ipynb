# Python OOP Journey: Part 1 - Classes and Instances

In this notebook, I am documenting my understanding of the absolute fundamentals of Object-Oriented Programming (OOP) in Python, inspired by Corey Schafer's Python OOP Tutorial 1. 

We will use a pure Python scenario (a Movie Catalog) to demonstrate the core concepts without any external libraries.

# Core Concepts Explained:
*Class (The Blueprint):Think of a class as an architectural blueprint or a mold. Writing a class doesn't create any data in memory; it just defines the structure.

*Instance (The Object):
An instance is the actual object built from that blueprint. Each instance contains its own unique data and is completely isolated in memory.

```python
# 1. Defining the blueprint (Class)
class Movie:
    """
    A blueprint that defines what a Movie object should look like and do.
    """
    
    # 2. The Constructor Method (__init__)
    # This runs AUTOMATICALLY every time we create a new movie.
    # 'self' is a mandatory parameter that represents the specific object being created.
    def __init__(self, title: str, director: str, release_year: int, rating: float):
        
        # Mapping the inputs to instance variables (Attributes)
        # Using 'self.' ensures these variables belong to this specific movie instance.
        self.title = title
        self.director = director
        self.release_year = release_year
        self.rating = rating

    # 3. Instance Method 1: To get a formatted description
    # It must take 'self' as an argument to access the instance's unique attributes.
    def get_description(self) -> str:
        return f"'{self.title}' directed by {self.director} ({self.release_year})"

    # 4. Instance Method 2: A simple conditional check
    # Returns True if the movie is highly rated, otherwise False.
    def is_highly_rated(self) -> bool:
        return self.rating >= 8.5
```

### Deep Dive: Understanding '__init__' and 'self'

* **Why do we need `__init__`?** The `__init__` method allows us to pass all the data cleanly in a single line when creating an object, instead of assigning attributes manually one by one.
* **What is the secret of `self`?** It acts as a pointer telling Python exactly which object's data to read or modify.
* When we call `movie_1.get_description()`, Python automatically passes `movie_1` as the first argument (`self`) behind the scenes.

```python
# --- Creating Unique Instances (Objects) in Memory ---
movie_1 = Movie(title="Heat", director="Michael Mann", release_year=1995, rating=8.3)
movie_2 = Movie(title="The Godfather", director="Francis Ford Coppola", release_year=1972, rating=9.2)

# --- Accessing Attributes Directly ---
print("--- Accessing Attributes Directly ---")
print(f"First Movie Title: {movie_1.title}")
print(f"Second Movie Director: {movie_2.director}")

# --- Calling Instance Methods ---
print("\n--- Calling Instance Methods ---")
print(movie_1.get_description())
print(movie_2.get_description())

# Testing the conditional logic method
print(f"Is '{movie_1.title}' highly rated? {movie_1.is_highly_rated()}")
print(f"Is '{movie_2.title}' highly rated? {movie_2.is_highly_rated()}")
```

### Summary and Conclusion

By building this notebook, I have successfully demonstrated:
1. How a **Class** acts as an isolated layout while **Instances** hold the real data.
2. How the **Constructor (`__init__`)** simplifies object initialization.
3. How **`self`** seamlessly connects methods to their corresponding object attributes.
